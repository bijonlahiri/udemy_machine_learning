```python
corpus = """Hello Welcome, to Bijon's course.
Please do watch the entire course! to become an expert in NLP.
"""

# corpus="""Short answer: yes… but only parts of it 🙂
# The ~/Library folder is a bit of a wild jungle on macOS, and the Python subfolders in there are usually support files, not the Python interpreter itself.
# Let’s break it down safely.
# """
```

### Tokenization
Convert paragraphs to sentences using NLTK

```python
# import nltk
# nltk.download()
```

```python
from nltk import sent_tokenize
sent_tokenize(corpus)
```

```python
## Convert a paragraph/sentence into words.
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
