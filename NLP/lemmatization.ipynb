## Lemmatization
Lemmatization is like stemming. The output of lemmatization is a **lemma**. It is a root word and not a word stem.
It does not change the meaning of the word.

### WordNet Lemmatizer
#### Post Tag for lemmatizer
- Noun - ('n')
- Verb - ('v')
- Adjective - ('a')
- Adverb - ('r')
- Satellite Adjectives - ('s')

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize('going', pos='n')
```

```python
lemmatizer.lemmatize('going', pos='v')
```
