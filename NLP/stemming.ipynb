## Stemming
It is a process to find out the root word for a given word.

### Porter Stemmer

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ['eating', 'eats', 'eaten', 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalize']
for word in words:
    print(f"{word}---->{stemmer.stem(word)}")
```

### Snowball Stemmer
It is better than Porter Stemmer

```python
from nltk.stem import SnowballStemmer
sb_stemmer = SnowballStemmer(language='english')
for word in words:
    print(f"{word}----------->{sb_stemmer.stem(word)}")
```

```python
stemmer.stem('Fairly'), stemmer.stem('sportingly')
```

```python
sb_stemmer.stem('fairly'), sb_stemmer.stem('sportingly')
```
